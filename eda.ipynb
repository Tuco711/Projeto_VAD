# Covid Dataset

```python
import pandas as pd
import plotly.io as pio
import numpy as np

pio.renderers.default = 'vscode'

try:
    df = pd.read_csv('datasets/owid-covid-data.csv')

except FileNotFoundError:
    df = pd.read_csv('/content/drive/MyDrive/Universidade/Mestrado/2 Semestre/Visualização e Análise de Dados/Projeto/Projeto_VAD/datasets/owid-covid-data.csv')

# Criar variaveis de tempo
df['date'] = pd.to_datetime(df['date'])

# Criar as colunas de periodicidade
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.to_period('M').dt.to_timestamp()
df['quarter'] = df['date'].dt.to_period('Q').dt.to_timestamp()

# Criar Semestre (S1 para meses 1-6, S2 para meses 7-12)
df['semester'] = np.where(df['date'].dt.month <= 6, 
                          df['date'].dt.year.astype(str) + '-01-01', 
                          df['date'].dt.year.astype(str) + '-07-01')
df['semester'] = pd.to_datetime(df['semester'])

df.head()
```

# Basic Plots

```python
import plotly.express as px
import plotly.graph_objects as go

df['date'] = pd.to_datetime(df['date'])

df_world = df[df['location'] == 'World']

fig = go.Figure()

# Linha de Casos (Eixo Y1)
fig.add_trace(go.Scatter(
    x=df_world['date'], 
    y=df_world['total_cases'],
    mode='lines',
    name='Total Cases',
    line=dict(color='red')
))

# Linha de Vacinações (Eixo Y2 - à direita)
fig.add_trace(go.Scatter(
    x=df_world['date'], 
    y=df_world['total_vaccinations'],
    mode='lines',
    name='Total Vaccinations',
    line=dict(color='blue'),
    yaxis='y2' # Define que esta linha usa o segundo eixo
))

# 4. Configurar o layout para suportar dois eixos
fig.update_layout(
    title='Evolução Global: Casos vs Vacinação (COVID-19)',
    xaxis_title='Data',
    yaxis=dict(
        title='Total de Casos',
        titlefont=dict(color='red'),
        tickfont=dict(color='red')
    ),
    yaxis2=dict(
        title='Total de Vacinações',
        titlefont=dict(color='blue'),
        tickfont=dict(color='blue'),
        overlaying='y', # Um eixo sobre o outro
        side='right'    # Eixo de vacinas à direita
    ),
    xaxis=dict(
        tickformat='%m-%Y', # Formato de data para o eixo X
        dtick='M6'         # Intervalo de 6 meses para os ticks do eixo X
    ),
    legend=dict(x=0.1, y=1.1, orientation='h')
)

fig.show()
```

```python
import pandas as pd
import plotly.graph_objects as go



# 1. Filtramos o Mundo ou um País para evitar duplicatas
df_plot = df[df['location'] == 'World'].copy()

# 2. Criando as métricas em Porcentagem (%)
# Nota: Usamos 'people_vaccinated' para saber quantas pessoas de fato receberam a vacina
df_plot['cases_pct'] = (df_plot['total_cases'] / df_plot['population']) * 100
df_plot['vac_pct'] = (df_plot['people_vaccinated'] / df_plot['population']) * 100

# 3. Gráfico
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=df_plot['date'], 
    y=df_plot['cases_pct'],
    mode='lines',
    name='% da População com COVID',
    line=dict(color='red')
))

fig.add_trace(go.Scatter(
    x=df_plot['date'], 
    y=df_plot['vac_pct'],
    mode='lines',
    name='% da População Vacinada (1ª dose)',
    line=dict(color='blue')
))

fig.update_layout(
    title='Impacto da COVID-19 vs Alcance da Vacinação (Escala Normalizada %)',
    xaxis_title='Data',
    yaxis_title='Porcentagem da População (%)',
    hovermode="x unified"
)

fig.show()
```
